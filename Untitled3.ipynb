```python
ans = []
n_estimators_values = []

# Loop through a range of n_estimators values to observe performance
# Starting from 10 to avoid the InvalidParameterError (n_estimators must be >= 1)
for i in range(10, 201, 10):
  model = RandomForestClassifier(n_estimators=i, random_state=42) # Initialize model with current n_estimators
  model.fit(X_train, y_train)
  y_pred = model.predict(X_test)
  from sklearn.metrics import accuracy_score
  accuracy = accuracy_score(y_test, y_pred)
  ans.append(accuracy)
  n_estimators_values.append(i)

# Plot the results after the loop
plt.figure(figsize=(10, 6))
plt.plot(n_estimators_values, ans, marker='o', linestyle='-')
plt.title('RandomForestClassifier Accuracy vs. n_estimators')
plt.xlabel('Number of Estimators (n_estimators)')
plt.ylabel('Accuracy Score')
plt.grid(True)
plt.xticks(n_estimators_values) # Show all n_estimators values on x-axis
plt.tight_layout()
plt.show()
```
